
# TF-IDF + Régression Logistique


## 1. Problématique

Les modèles de Machine Learning ne comprennent que des nombres. 
Le texte doit donc être transformé en représentation numérique. Pour cela, on utilise souvent la technique TF-IDF (Term Frequency - Inverse Document Frequency).



## 2. TF-IDF

Le but c'est de représenter chaque document par un vecteur numérique qui reflète l'importance de chaque mot dans le document par rapport à l'ensemble du corpus.

### TF (Term Frequency)
$$TF(t,d) = \frac{\text{occurrences de } t}{\text{nb total de mots dans } d}$$

### IDF (Inverse Document Frequency)
$$IDF(t) = \log \left( \frac{N}{df(t)} \right)$$
Avec :
- N : nombre total de documents
- df(t) : nombre de documents contenant le terme t

### TF-IDF
$$TF\text{-}IDF(t,d) = TF(t,d) \times IDF(t)$$

Pour chaque document, on se retrouve avec un vecteur numérique du style [0.0, 0.42, 0.0, 0.91, ...] qui correspond à l'importance de chaque mot du vocabulaire dans le document.


## 3. Régression Logistique

Une fois qu'on a notre vecteur TF-IDF, on peut l'utiliser comme entrée pour une **régression logistique** afin de classer les documents.

### Formule :
$$z = w \cdot x + b$$

### Sigmoïde :
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

Donne une **probabilité de classe**



## 4. Fonction de coût

**Log-loss (Cross Entropy)** :
$$L(y, \hat{y}) = -[y \log(\hat{y}) + (1-y)\log(1-\hat{y})]$$

Objectif : minimiser l'erreur de classification.



## 5. Exemple


```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        ngram_range=(1, 2),
        max_df=0.9,
        min_df=5,
        stop_words="english"
    )),
    ("clf", LogisticRegression(max_iter=1000))
])

# Exemple d'entraînement
X = ["this is a good movie", "this is a bad movie"]
y = [1, 0]

pipeline.fit(X, y)
pipeline.predict(["a good movie"])
```


## 6. Hyperparamètres importants

### TF-IDF
- `ngram_range` : taille des n-grammes
- `min_df`, `max_df` : fréquence minimale et maximale des mots
- `stop_words` : liste des mots à ignorer
